## 3.A — Validar coordenadas geográficas

Detecta registros donde `decimalLatitude` no esté en `[-90, 90]`, `decimalLongitude` no esté en `[-180, 180]`, o alguno de los valores no pueda convertirse a número.

```python
cantidad, errores = validaciones.validar_coordenadas_incompletas(database)
print(f"Registros con coordenadas incompletas: {cantidad}")
for e in errores[:10]:
    print(e)
```

## 3.C — Validar fechas

Detecta fechas con formato inválido o posteriores al año actual. 
Consideramos como válidos los formatos de fechas que cumplan con los formatos que define el estándar Darwin Core y GBIF. 

```python

cantidad, errores = validaciones.verificar_fechas(database)
print(f"\n=== {database} ===")
print(f"Cantidad de errores: {cantidad}")
for error in errores[:10] :#solo los primeros 10
        print(error)
```
