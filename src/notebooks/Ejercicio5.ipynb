# Ejercicio 5 — Actualización de registros en datasets

> **Objetivo:** Identificar registros específicos dentro de un dataset, modificar uno o más de sus campos y generar una nueva versión del dataset con los cambios aplicados.
>
>Los archivos originales en `raw_datasets` **no se modifican**. Toda actualización genera un nuevo archivo en `processed_datasets`.

````python
import os
import sys

# os.path.pardir sube de 'notebooks' a 'src'
ruta = os.path.abspath(os.path.pardir)
if ruta not in sys.path:
    sys.path.append(ruta)

import core.ejercicio5_actualizacion_registros as actualizacion

# Dataset con el que trabajamos en este notebook
# countryCode: AR | occurrenceID formato: IADIZA:COI:XXXXXX
DATASET = 'IADIZA'
````

---
## 5.A — Búsqueda de registros por múltiples columnas

Función que filtra registros por uno o más campos simultáneamente. Retorna todos los campos de los registros que cumplen la condición.

````python
# Búsqueda por occurrenceID
resultados = actualizacion.buscar_registros(DATASET, {'occurrenceID': 'IADIZA:COI:006861'})
print(f"Registros encontrados: {len(resultados)}")
for r in resultados:
    print(r)
````

````python
# Búsqueda por scientificName + countryCode
resultados = actualizacion.buscar_registros(DATASET, {'scientificName': 'Chloephaga picta (J.F.Gmelin, 1789)', 'countryCode': 'AR'})
print(f"Registros encontrados: {len(resultados)}")
for r in resultados:
    print(r)
````

````python
# Búsqueda solo por countryCode
resultados = actualizacion.buscar_registros(DATASET, {'countryCode': 'AR'})
print(f"Registros encontrados: {len(resultados)}")
for r in resultados[:5]:
    print(r)
````

---
## 5.B — Actualizar un campo de un registro

Dado el identificador de un registro, el nombre de una columna y un nuevo valor, actualiza ese campo en particular y genera un nuevo archivo en `processed_datasets`.

````python
# Antes de actualizar
antes = actualizacion.buscar_registros(DATASET, {'occurrenceID': 'IADIZA:COI:006861'})
print("Antes:", antes[0].get('stateProvince'))

# Actualizamos el campo 'stateProvince' del registro IADIZA:COI:006861
resultado = actualizacion.actualizar_campo_registro(
    DATASET,
    'occurrenceID',       # columna identificadora
    'IADIZA:COI:006861',  # valor que identifica el registro
    'stateProvince',      # campo a modificar
    'Mendoza'             # nuevo valor
)
print(resultado)

# Después de actualizar
despues = actualizacion.buscar_registros(DATASET, {'occurrenceID': 'IADIZA:COI:006861'})
print("Después:", despues[0].get('stateProvince'))
````

---
## 5.C — Actualizar múltiples campos de un registro

Permite modificar varios campos en una sola operación, recibiendo un diccionario con los cambios a aplicar.

```python
{'countryCode': 'AR', 'stateProvince': 'Buenos Aires'}
```

````python
# Antes de actualizar
antes = actualizacion.buscar_registros(DATASET, {'occurrenceID': 'IADIZA:COI:006861'})
print("Antes - countryCode:", antes[0].get('countryCode'))
print("Antes - stateProvince:", antes[0].get('stateProvince'))

# Actualizamos countryCode y stateProvince del mismo registro
nuevos_valores = {
    'countryCode': 'BR',
    'stateProvince': 'Mendoza'
}

resultado = actualizacion.actualizar_multiples_campos(
    DATASET,
    'occurrenceID',
    'IADIZA:COI:006861',
    nuevos_valores
)
print(resultado)

# Después de actualizar
despues = actualizacion.buscar_registros(DATASET, {'occurrenceID': 'IADIZA:COI:006861'})
print("Después - countryCode:", despues[0].get('countryCode'))
print("Después - stateProvince:", despues[0].get('stateProvince'))
````

## 5.D — Actualización con validación previa

Antes de aplicar cambios, valida que los nuevos valores sean correctos reutilizando las funciones del Ejercicio 3. Si la validación falla, el registro **no se modifica** y se reportan los errores.

````python
# Caso inválido: countryCode inexistente → debe rechazarse
# (el dataset IADIZA no tiene coordenadas en ningún registro, entonces la validación siempre detecta eso)
exito, mensaje = actualizacion.actualizar_registro_validado(
    DATASET,
    'occurrenceID',
    'IADIZA:COI:006861',
    'countryCode',
    'ZZ'  # código de país inválido
)
print(f"{'válido' if exito else 'inválido'} {mensaje}")
````
